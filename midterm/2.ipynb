```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd

# Шаг 1: Сбор данных
df = pd.read_csv("Companies_ranked_by_Dividend_Yield.csv")

# Выводим первые строки и информацию о данных
print(df.head())
print(df.info())

# Шаг 2: Очистка данных и обработка пропущенных значений
# Удаляем строки с пропущенными значениями
df_cleaned = df.dropna()

# Шаг 3: Кодирование категориальных переменных
# Используем OneHotEncoder для колонки 'country'
encoder = OneHotEncoder(drop='first', sparse_output=False)  # Use sparse_output instead of sparse
encoded_country = encoder.fit_transform(df_cleaned[['country']])

# Создаем DataFrame с закодированными странами
encoded_country_df = pd.DataFrame(encoded_country, columns=encoder.get_feature_names_out(['country']))

# Шаг 4: Масштабирование числовых признаков
# Масштабируем числовые переменные 'dividend_yield_ttm' и 'price (GBP)'
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df_cleaned[['dividend_yield_ttm', 'price (GBP)']])

# Создаем DataFrame с масштабированными данными
scaled_features_df = pd.DataFrame(scaled_features, columns=['dividend_yield_ttm_scaled', 'price_scaled'])

# Шаг 5: Объединение обработанных данных
# Объединяем исходные данные с закодированными и масштабированными признаками
df_processed = pd.concat([df_cleaned[['Rank', 'Name', 'Symbol']], encoded_country_df, scaled_features_df], axis=1)

# Выводим обработанный DataFrame
print(df_processed.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Загрузим данные
df_cleaned = pd.read_csv('Companies_ranked_by_Dividend_Yield.csv').dropna()

# Проверка данных
print(df_cleaned.head())
print(df_cleaned.columns)

# 2. Корреляция между числовыми признаками
correlation_matrix = df_cleaned[['dividend_yield_ttm', 'price (GBP)']].corr()
print(correlation_matrix)

# Визуализация корреляционной матрицы
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()

# 3. Масштабирование данных перед PCA
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df_cleaned[['dividend_yield_ttm', 'price (GBP)']])

# Применение PCA для снижения размерности
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

# Визуализация результатов PCA
plt.figure(figsize=(8, 6))
plt.scatter(pca_result[:, 0], pca_result[:, 1], alpha=0.5)
plt.title('PCA: Dimensionality Reduction')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

# Вывод объяснённой дисперсии
print('Explained variance ratio:', pca.explained_variance_ratio_)
print('Total explained variance:', np.sum(pca.explained_variance_ratio_))
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

# 1. Загрузка и предварительная обработка данных
df_cleaned = pd.read_csv('Companies_ranked_by_Dividend_Yield.csv').dropna()

# Предположим, что целевой колонкой будет 'country' (или укажи другую, если это не так)
# Преобразуем категориальные значения целевой переменной в числовые
label_encoder = LabelEncoder()
df_cleaned['country_encoded'] = label_encoder.fit_transform(df_cleaned['country'])

# 2. Разделим данные на признаки и целевую переменную
X = df_cleaned[['dividend_yield_ttm', 'price (GBP)']]  # Признаки для классификации
y = df_cleaned['country_encoded']  # Целевая переменная

# Масштабируем данные (важно для логистической регрессии)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Разделение данных на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# 4. Применение логистической регрессии
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
y_pred_logreg = logreg.predict(X_test)

# 5. Применение дерева решений
tree = DecisionTreeClassifier()
tree.fit(X_train, y_train)
y_pred_tree = tree.predict(X_test)

# 6. Оценка моделей
print("Logistic Regression Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_logreg)}")
print(f"Precision: {precision_score(y_test, y_pred_logreg, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_logreg, average='weighted')}")
print("\nClassification Report for Logistic Regression:")
print(classification_report(y_test, y_pred_logreg))

print("\nDecision Tree Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_tree)}")
print(f"Precision: {precision_score(y_test, y_pred_tree, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_tree, average='weighted')}")
print("\nClassification Report for Decision Tree:")
print(classification_report(y_test, y_pred_tree))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

# Данные уже разделены на X_train, X_test, y_train, y_test из предыдущих шагов

# 1. Применение Random Forest
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

# 2. Применение Support Vector Machine (SVM)
svm = SVC()
svm.fit(X_train, y_train)
y_pred_svm = svm.predict(X_test)

# 3. Оценка моделей: Random Forest
print("\nRandom Forest Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf)}")
print(f"Precision: {precision_score(y_test, y_pred_rf, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_rf, average='weighted')}")
print("\nClassification Report for Random Forest:")
print(classification_report(y_test, y_pred_rf))

# 4. Оценка моделей: SVM
print("\nSupport Vector Machine (SVM) Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm)}")
print(f"Precision: {precision_score(y_test, y_pred_svm, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_svm, average='weighted')}")
print("\nClassification Report for SVM:")
print(classification_report(y_test, y_pred_svm))

# Теперь сравним с базовыми моделями (логистическая регрессия и дерево решений)
print("\nLogistic Regression Performance (from previous step):")
print(f"Accuracy: {accuracy_score(y_test, y_pred_logreg)}")
print(f"Precision: {precision_score(y_test, y_pred_logreg, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_logreg, average='weighted')}")
print("\nClassification Report for Logistic Regression:")
print(classification_report(y_test, y_pred_logreg))

print("\nDecision Tree Performance (from previous step):")
print(f"Accuracy: {accuracy_score(y_test, y_pred_tree)}")
print(f"Precision: {precision_score(y_test, y_pred_tree, average='weighted')}")
print(f"Recall: {recall_score(y_test, y_pred_tree, average='weighted')}")
print("\nClassification Report for Decision Tree:")
print(classification_report(y_test, y_pred_tree))

# Здесь мы можем определить модель с лучшими показателями по accuracy, precision, и recall.
```
